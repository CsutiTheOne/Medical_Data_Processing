```python
def predictionsPerClass(CM, className):
    TP = CM[className][className]
    FN = CM[className].sum() - TP
    FP = CM.loc[className].sum() - TP
    TN = CM.values.sum() - (TP + FN + FP)
    return TP, FN, FP, TN

def classMetrics(TP, FN, FP, TN):
    try:
        accuracy = (TP + TN) / (TP + FN + FP + TN)
        precision = TP / (TP + FP)
        recall = TP / (TP + FN)
        F1_score = (2*precision*recall) / (precision + recall)
        return accuracy, precision, recall, F1_score
    except:
        return 0,0,0,0
```

```python
import os
from Training.consts import DS_PATH

neviRoot = f"{DS_PATH}/test/Melanocytic nevi/"
neviPaths = [ neviRoot + fn for fn in os.listdir(neviRoot) ]

melanomaRoot = f"{DS_PATH}/test/Melanoma/"
melanomaPaths = [ melanomaRoot + fn for fn in os.listdir(melanomaRoot) ]


CoAtNetPreds_nevi = []
CoAtNetPreds_melanoma = []
with open("CoAtNetPreds.txt", "r") as f:
    lines = f.readlines()
    for i, l in enumerate(lines):
        label, pred = l.strip().split(",")
        label = int(label)
        pred = int(pred.strip())
        if label == 4:
            CoAtNetPreds_nevi.append((label, pred))
        if label == 5:
            CoAtNetPreds_melanoma.append((label, pred))

SwinPreds_nevi = []
SwinPreds_melanoma = []
with open("SwinPreds.txt", "r") as f:
    lines = f.readlines()
    for l in lines:
        label, pred = l.strip().split(",")
        label = int(label)
        pred = int(pred.strip())
        if label == 4:
            SwinPreds_nevi.append((label, pred))
        if label == 5:
            SwinPreds_melanoma.append((label, pred))
            
```

```python
preds = {}

for i in range(len(neviPaths)):
    preds[neviPaths[i]] = {
        "label": SwinPreds_nevi[i][0],
        "Swin": SwinPreds_nevi[i][1],
        "CoAtNet": CoAtNetPreds_nevi[i][1]
    }

for i in range(len(melanomaPaths)):
    preds[melanomaPaths[i]] = {
        "label": SwinPreds_melanoma[i][0],
        "Swin": SwinPreds_melanoma[i][1],
        "CoAtNet": CoAtNetPreds_melanoma[i][1]
    }
```
